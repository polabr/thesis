```python
import ROOT as rt
from ROOT import TFile, TTree, TCanvas, TH2D, TH1F
```

```python
h_trueAndReco = rt.TH1F("h_trueAndReco","Purity",100, 0, 2)
#h_trueAndReco.SetFillColor(rt.kRed)
h_true = rt.TH1F("h_true","True",100, 0, 2)
#h_true.SetFillColor(rt.kBlue)
h_reco = rt.TH1F("h_reco","Reco",100, 0, 2)
h_reco.SetFillColor(rt.kBlue)

h_reco_cosmics = rt.TH1F("h_reco_cosmics","Reco Cosmics",100, 0, 2)
h_reco_cosmics.SetFillColor(rt.kRed)
```

```python
#f = TFile("selectedEventsTrueRecoBoth_vtxOnly_121824.root","READ") # vtx cut only
#f = TFile("selectedEventsTrueRecoBoth_muOnly_121824.root","READ") # mu cut only
f = TFile("selectedEventsTrueRecoBoth_withTrueTrkDir_withParticleEs_111124.root","READ")
t = f.Get("selectedEvents")

#f2 = TFile("selectedEventsTrueRecoBoth_cosmics_vtxOnly_fixed_011325.root","READ") # vtx cut only
#f2 = TFile("selectedEventsTrueRecoBoth_cosmics_muOnly_121824.root","READ") # mu cut only
f2 = TFile("selectedEventsTrueRecoBoth_cosmics_fixed_120224.root","READ")
t2 = f2.Get("selectedEvents")


entries = t.GetEntries()
print(entries)

entries2 = t2.GetEntries()
print(entries2)
```

```python
# making plots in terms of reco neutrino energy on the x-axis (so filling with this)

# for calculating the total purity later
sum_overlay_weights_both_list = []
sum_overlay_weights_selected_list = []

for e in range(entries):

    t.GetEntry(e)
    
    if (t.passedSel_ == 1) or (t.passedSel_ == 3): # passes truth selection
        h_true.Fill(t.recoSel_recoNuE_, t.weight_)
        if (t.passedSel_ == 2) or (t.passedSel_ == 3): # passes reco selection as well (both selections)
            h_trueAndReco.Fill(t.recoSel_recoNuE_, t.weight_)
            sum_overlay_weights_both_list.append(t.weight_)
    if (t.passedSel_ == 2) or (t.passedSel_ == 3): # passes reco selection
        #print("This entry passed the reco sel.")
        #print("The value is: ", t.recoSel_recoNuE_)
        h_reco.Fill(t.recoSel_recoNuE_, t.weight_)
        sum_overlay_weights_selected_list.append(t.weight_)
```

```python
# cosmics looop

for e in range(entries2):

    t2.GetEntry(e)
    
    #if (t2.passedSel_ == 1) or (t2.passedSel_ == 3): # passes truth selection
    #    h_true.Fill(t.recoSel_recoMomMu_, t.weight_)
    #    if (t.passedSel_ == 2) or (t.passedSel_ == 3): # passes reco selection as well (both selections)
    #        h_trueAndReco.Fill(t.recoSel_recoMomMu_, t.weight_)
    if (t2.passedSel_ == 2) or (t2.passedSel_ == 3): # passes reco selection
        #print("This entry passed the reco sel.")
        #print("The value is: ", t2.recoSel_recoNuE_)
        h_reco_cosmics.Fill(t2.recoSel_recoNuE_)
```

```python
targetPOT = 1.3e21 # total expected POT for runs 1-5 in MicroBooNE
simPOT = 4.68e20 # this is for run 1
simPOT_cosmics = 3.297e20 # this is for run 1 EXTBNB

h_reco.Scale(targetPOT/simPOT)
h_reco_cosmics.Scale(targetPOT/simPOT_cosmics)
```

```python
# calculate total purity of this sample
selected_signal = sum(sum_overlay_weights_both_list)*(targetPOT/simPOT)
all_selected = sum(sum_overlay_weights_selected_list)*(targetPOT/simPOT) + (h_reco_cosmics.GetEntries())*(targetPOT/simPOT_cosmics)
print(selected_signal, ", ", all_selected)
print("purity: ", (selected_signal / all_selected)*100, "%")
```
